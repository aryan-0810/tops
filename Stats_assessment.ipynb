Q.1 Convert a 1D array to a 2D array with 2 rows

```python
import numpy as np

a = np.array([1, 2, 3, 4, 5, 6, 7, 8])
result = a.reshape(2, -1)

print(result)
```

Q.2 Get the common items between a and b

```python
a = np.array([1, 2, 3, 2, 3, 4, 3, 4, 5, 6])
b = np.array([7, 2, 10, 2, 7, 4, 9, 4, 9, 8])

result = np.intersect1d(a, b)

print(result)
```

Q.3 Get all items between 5 and 10 from a

```python
a = np.array([2, 6, 1, 9, 10, 3, 27])

result = a[(a >= 5) & (a <= 10)]

print(result)
```

Q.10 Create an array of numbers between 0 and 6 with increments of 0.3 and name it x. On the same plot, plot x, x², x³, and x⁴

```python
x = np.arange(0, 6.1, 0.3)

plt.plot(x, x, "ro")
plt.plot(x, x**2, "bs")
plt.plot(x, x**3, "g")
plt.plot(x, x**4, ":")

plt.xlim(0, 6)
plt.ylim(0, 125)

plt.show()
```

TENSORFLOW AND MACHINE LEARNING

Q.13 What are the features of TensorFlow?

TensorFlow is an open-source machine learning and deep learning framework developed by Google

Open Source: TensorFlow is freely available and has a large developer community.
Deep Learning Support: It can be used to build neural networks such as CNNs, RNNs, and Transformers.
GPU and TPU Support: It can use GPUs and TPUs to accelerate machine learning and deep learning computations.
Scalability: TensorFlow can run on personal computers as well as large distributed systems.
Automatic Differentiation: It automatically calculates gradients required for training neural networks.
Keras Integration: TensorFlow provides the Keras API for easily building and training deep learning models.
Model Deployment: Trained models can be deployed on web applications, mobile devices, servers, and edge devices.
Tensor Operations: It efficiently performs mathematical operations using multidimensional arrays called tensors.
Visualization: TensorBoard can be used to visualize training metrics, model performance, and computational graphs.
Large Community: TensorFlow has extensive documentation, tutorials, and community support.

Q.14 List a few limitations of TensorFlow

Steep Learning Curve: Beginners may find TensorFlow difficult to learn.
Complex Syntax: Some operations can require more code compared with simpler machine learning libraries.
High Resource Requirements: Large deep learning models may require significant CPU, GPU, RAM, and storage resources.
Debugging Difficulty: Debugging complex neural networks can sometimes be challenging.
Overkill for Simple Tasks: TensorFlow may be unnecessary for simple machine learning tasks where libraries like Scikit-learn are easier to use.
Large Installation Size: TensorFlow and its dependencies can require considerable disk space.
Hardware Dependency: Training large models efficiently may require powerful GPUs or TPUs.
Long Training Time: Complex deep learning models can take a long time to train.

Q.15 What do you know about supervised and unsupervised machine learning?

Supervised Machine Learning

Supervised learning is a type of machine learning where the model is trained using labeled data. This means that the input data contains a known target or output.

The model learns the relationship between the input features and the target variable and then uses this knowledge to make predictions on new data.

The two main types of supervised learning are:

1. Regression: Used when the target variable is continuous or numerical.

Examples:

Predicting house prices
Predicting car prices
Predicting salary

Algorithms:

Linear Regression
Decision Tree Regression
Random Forest Regression
Support Vector Regression

2. Classification: Used when the target variable consists of categories or classes.

Examples:

Spam or Not Spam
Disease or No Disease
Pass or Fail

Algorithms:

Logistic Regression
Decision Tree
Random Forest
K-Nearest Neighbors
Naive Bayes
Support Vector Machine
Unsupervised Machine Learning

Unsupervised learning is a type of machine learning where the model is trained using unlabeled data. There is no predefined target variable. The algorithm tries to find hidden patterns, structures, or relationships within the data.

Common types of unsupervised learning include:

1. Clustering: Groups similar data points together.

Examples:

Customer segmentation
Grouping similar products
Market segmentation

Algorithms:

K-Means Clustering
Hierarchical Clustering
DBSCAN

2. Dimensionality Reduction: Reduces the number of features while preserving important information.

Example:

PCA (Principal Component Analysis)
Difference Between Supervised and Unsupervised Learning
Feature	Supervised Learning	Unsupervised Learning
Data	Labeled data	Unlabeled data
Target Variable	Present	Not present
Main Purpose	Prediction	Finding patterns
Common Tasks	Classification, Regression	Clustering, Dimensionality Reduction
Example	Predict house price	Group customers
Algorithms	Linear Regression, Random Forest, SVM	K-Means, DBSCAN, PCA
